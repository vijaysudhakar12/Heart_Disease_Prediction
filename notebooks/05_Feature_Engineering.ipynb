Encode the categorical features using OneHotEncoder

```python
onehot_encoder = OneHotEncoder(sparse_output=False, drop='first')

def one_encode(df, col):
    encoded = onehot_encoder.fit_transform(df[[col]])

    encoder_df = pd.DataFrame(encoded, columns=onehot_encoder.get_feature_names_out())

    df = df.reset_index(drop=True)

    df = pd.concat([df, encoder_df], axis=1)

    df = df.drop(columns=col)

    return df
```

```python
from sklearn.preprocessing import OrdinalEncoder

ordinal_encoder = OrdinalEncoder()

def ordinal_encode(df, col):
    encoded = ordinal_encoder.fit_transform(df[[col]])

    encoder_df = pd.DataFrame(encoded, columns=ordinal_encoder.get_feature_names_out())

    df = df.reset_index(drop=True)

    df = pd.concat([df.drop(columns=col), encoder_df], axis=1)

    return df
```
